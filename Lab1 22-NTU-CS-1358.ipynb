***Task1-Loading dataset and showing features***

```python
#Import Libraries
from sklearn.datasets import load_iris
iris = load_iris()
print(iris.keys)
print(iris["target_names"])
```

```python
n_samples,n_features = iris.data.shape
print('Number of Samples:', n_samples)
print('Number of Features:', n_features)    
```

```python
print('Feature Name:', iris.feature_names)
print('Target Name:', iris.target_names)
```

```python
print('Dimension of Input', iris.data.shape)
print('Dimension of Output', iris.target.shape)
```

```python
print("First 5 rows of Input:", iris.data[1:5])
print("Target Data:", iris.target)
```

***Task2-Histogram of the features***

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots()
x_index = 3
colors = ['red', 'black', 'yellow']
for label, color in zip(range(len(iris.target_names)),colors):
    ax.hist(iris.data[iris.target==label, x_index], label=iris.target_names[label], color=color)
    ax.set_xlabel(iris.feature_names[x_index])
    ax.legend(loc='upper right')
fig.show()
```

***Task3-Scattered plot with two features***

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
iris = load_iris()
fig, ax = plt.subplots()
x_index = 3
y_index = 0
colors = ['black', 'blue', 'green']
for label, color in zip(range(len(iris.target_names)), colors):
    ax.scatter(iris.data[iris.target == label, x_index], iris.data[iris.target==label, y_index],label=iris.target_names[label],c=color)
    ax.set_xlabel(iris.feature_names[x_index])
    ax.set_ylabel(iris.feature_names[y_index])
    ax.legend(loc='upper left')
```

***Task4-Scatter Plot Total***

```python
import matplotlib.pyplot as plt
n = len(iris.feature_names)
fig,ax = plt.subplots(n, n, figsize = (16,16))
colors = ['black', 'blue', 'green']
for x in range(n):
    for y in range(n):
        xname = iris.feature_names[x]
        yname = iris.feature_names[y]
        for color_ind in range(len(iris.target_names)):
            ax[x,y].scatter(iris.data[iris.target==color_ind, x], iris.data[iris.target==color_ind,y], label=iris.target_names[color_ind], c=colors[color_ind])
        ax[x,y]. set_xlabel(xname)
        ax[x,y]. set_ylabel(yname)
        ax[x,y].legend(loc='upper right')
plt.show()        
```
